```python
# Simple logic program

# 1. Basic logic with and / or / not
a = True
b = False

print("a and b:", a and b)
print("a or b:", a or b)
print("not a:", not a)

# 2. Take input and make decisions
age = int(input("Enter your age: "))

if age >= 18:
    print("You are an adult")
elif age >= 13:
    print("You are a teenager")
else:
    print("You are a child")

# 3. Check if a number is even or odd
number = int(input("Enter a number: "))

if number % 2 == 0:
    print(number, "is even")
else:
    print(number, "is odd")

# 4. Loop with a condition
for i in range(1, 6):
    if i % 2 == 0:
        print(i, "is even")
    else:
        print(i, "is odd")
```
