# “Series”
Fibonacci, arithmetic, and geometric series, plus the sum of the first n terms


```python
import math

def triangular(n):
    return [i * (i + 1) // 2 for i in range(1, n + 1)]   # 1, 3, 6, 10, ...

def factorials(n):
    return [math.factorial(i) for i in range(1, n + 1)]  # 1, 2, 6, 24, ...

def collatz(start):
    seq = [start]
    while start != 1:
        start = start // 2 if start % 2 == 0 else 3 * start + 1
        seq.append(start)
    return seq

def e_series(terms):
    # e = 1 + 1/1! + 1/2! + 1/3! + ...
    return sum(1 / math.factorial(k) for k in range(terms))

def star_pyramid(n):
    for i in range(1, n + 1):
        print(" " * (n - i) + "*" * (2 * i - 1))

def number_pattern(n):
    for i in range(1, n + 1):
        print("".join(str(j) for j in range(1, i + 1)))

def pascal(n):
    row = [1]
    rows = []
    for _ in range(n):
        rows.append(row)
        row = [1] + [row[i] + row[i + 1] for i in range(len(row) - 1)] + [1]
    width = len(" ".join(map(str, rows[-1])))
    for r in rows:
        print(" ".join(map(str, r)).center(width))

n = int(input("How many terms/rows? "))

print("Triangular:", triangular(n))
print("Factorial :", factorials(n))
print("Collatz(6):", collatz(6))
print(f"e by series ({n} terms): {e_series(n):.6f}  |  math.e: {math.e:.6f}")

print("\nStar pyramid:")
star_pyramid(n)

print("\nNumber pattern:")
number_pattern(n)

print("\nPascal's triangle:")
pascal(n)
```
