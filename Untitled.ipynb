GCD/LCM, prime factorization, a quadratic solver, Newton’s method, numerical integration and differentiation, matrix operations, Euler’s totient, and basic statistics.

```python
import math
import cmath
import statistics

def lcm(a, b):
    return a * b // math.gcd(a, b)

def prime_factors(n):
    factors, d = {}, 2
    while d * d <= n:
        while n % d == 0:
            factors[d] = factors.get(d, 0) + 1
            n //= d
        d += 1
    if n > 1:
        factors[n] = factors.get(n, 0) + 1
    return factors

def totient(n):
    # Euler's phi: count of numbers <= n coprime to n
    result = n
    for p in prime_factors(n):
        result -= result // p
    return result

def quadratic(a, b, c):
    # ax^2 + bx + c = 0 (handles complex roots too)
    d = cmath.sqrt(b * b - 4 * a * c)
    return (-b + d) / (2 * a), (-b - d) / (2 * a)

def newton_sqrt(x, iterations=20):
    # Newton's method on f(g) = g^2 - x
    g = x / 2
    for _ in range(iterations):
        g = g - (g * g - x) / (2 * g)
    return g

def derivative(f, x, h=1e-5):
    return (f(x + h) - f(x - h)) / (2 * h)        # central difference

def simpson(f, a, b, n=100):
    # n must be even
    h = (b - a) / n
    total = f(a) + f(b)
    for i in range(1, n):
        total += f(a + i * h) * (4 if i % 2 else 2)
    return total * h / 3

def mat_mul(A, B):
    return [[sum(A[i][k] * B[k][j] for k in range(len(B)))
             for j in range(len(B[0]))] for i in range(len(A))]

def det(M):
    # recursive cofactor expansion along the first row
    if len(M) == 1:
        return M[0][0]
    return sum((-1) ** j * M[0][j] *
               det([row[:j] + row[j + 1:] for row in M[1:]])
               for j in range(len(M)))

a, b = 48, 18
print(f"GCD({a},{b}) = {math.gcd(a, b)} | LCM({a},{b}) = {lcm(a, b)}")
print("Prime factors of 360:", prime_factors(360))
print("Euler totient of 36 :", totient(36))
print("Roots of x^2 - 3x + 2:", quadratic(1, -3, 2))
print("Roots of x^2 + 1     :", quadratic(1, 0, 1))
print(f"sqrt(2) by Newton    : {newton_sqrt(2):.6f}")
print(f"d/dx x^3 at x=2      : {derivative(lambda x: x ** 3, 2):.4f}")
print(f"Integral x^2, 0 to 3 : {simpson(lambda x: x ** 2, 0, 3):.4f}")
print(f"Integral sin, 0 to pi: {simpson(math.sin, 0, math.pi):.4f}")

A = [[1, 2], [3, 4]]
B = [[5, 6], [7, 8]]
print("A x B      :", mat_mul(A, B))
print("det(A)     :", det(A))
print("det(3x3)   :", det([[2, 0, 1], [1, 3, 2], [1, 1, 1]]))

print("C(10,3) =", math.comb(10, 3), "| P(10,3) =", math.perm(10, 3))

data = [2, 4, 4, 4, 5, 5, 7, 9]
print("Mean:", statistics.mean(data),
      "| Median:", statistics.median(data),
      "| Std dev:", statistics.pstdev(data))
```
