# “Series”
Pell, Catalan, Padovan, look-and-say, happy numbers, and the Basel and alternating harmonic series.


```python
import math

def pell(n):
    a, b = 0, 1
    out = []
    for _ in range(n):
        out.append(a)
        a, b = b, 2 * b + a
    return out

def catalan(n):
    return [math.comb(2 * i, i) // (i + 1) for i in range(n)]

def padovan(n):
    seq = [1, 1, 1]
    while len(seq) < n:
        seq.append(seq[-2] + seq[-3])
    return seq[:n]

def look_and_say(n):
    term, out = "1", []
    for _ in range(n):
        out.append(term)
        new, i = "", 0
        while i < len(term):
            j = i
            while j < len(term) and term[j] == term[i]:
                j += 1
            new += str(j - i) + term[i]
            i = j
        term = new
    return out

def is_happy(num):
    seen = set()
    while num != 1 and num not in seen:
        seen.add(num)
        num = sum(int(d) ** 2 for d in str(num))
    return num == 1

def basel(terms):
    # 1 + 1/2^2 + 1/3^2 + ...  -> pi^2 / 6
    return sum(1 / k ** 2 for k in range(1, terms + 1))

def alt_harmonic(terms):
    # 1 - 1/2 + 1/3 - 1/4 + ...  -> ln(2)
    return sum((-1) ** (k + 1) / k for k in range(1, terms + 1))

def infinite_gp(first, ratio):
    # valid only when |ratio| < 1
    return first / (1 - ratio)

n = int(input("How many terms? "))

print("Pell         :", pell(n))
print("Catalan      :", catalan(n))
print("Padovan      :", padovan(n))
print("Look-and-say :", look_and_say(n))
print("Happy numbers up to 50:", [i for i in range(1, 51) if is_happy(i)])

print(f"\nBasel ({n} terms)  : {basel(n):.6f} | pi^2/6 = {math.pi ** 2 / 6:.6f}")
print(f"ln(2) ({n} terms)  : {alt_harmonic(n):.6f} | math.log(2) = {math.log(2):.6f}")
print("Infinite GP 1 + 1/2 + 1/4 + ... =", infinite_gp(1, 0.5))
```
