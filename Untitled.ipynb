```python
from itertools import product

# 1. Basic logic gates
def AND(a, b):  return a and b
def OR(a, b):   return a or b
def NOT(a):     return not a
def XOR(a, b):  return a != b
def NAND(a, b): return not (a and b)
def NOR(a, b):  return not (a or b)

# 2. Truth table generator for any two-input gate
def truth_table(name, gate):
    print(f"\n{name} gate")
    print("A     B     Output")
    print("-" * 20)
    for a, b in product([False, True], repeat=2):
        print(f"{str(a):<6}{str(b):<6}{gate(a, b)}")

# 3. Decision logic with conditionals
def check_eligibility(age, has_id, is_member):
    if age >= 18 and has_id:
        if is_member:
            return "Access granted (member)"
        return "Access granted (guest)"
    elif age >= 18 and not has_id:
        return "Denied: ID required"
    else:
        return "Denied: must be 18 or older"

# 4. Half adder built from logic gates
def half_adder(a, b):
    total = XOR(a, b)
    carry = AND(a, b)
    return total, carry


if __name__ == "__main__":
    for name, gate in [("AND", AND), ("OR", OR), ("XOR", XOR),
                       ("NAND", NAND), ("NOR", NOR)]:
        truth_table(name, gate)

    print("\nEligibility checks:")
    print(check_eligibility(20, True, True))
    print(check_eligibility(20, False, False))
    print(check_eligibility(16, True, False))

    print("\nHalf adder (1 + 1):")
    s, c = half_adder(True, True)
    print(f"Sum = {int(s)}, Carry = {int(c)}")
```
