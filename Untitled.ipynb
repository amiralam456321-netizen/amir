Bisection method, Gaussian elimination, Lagrange interpolation, Euler’s method for differential equations, modular inverse, Chinese remainder theorem, fractions, linear regression, vector operations, and Fibonacci by matrix power.

```python
import math
from fractions import Fraction

def bisection(f, a, b, tol=1e-10):
    # finds a root of f in [a, b]; f(a) and f(b) must have opposite signs
    while b - a > tol:
        mid = (a + b) / 2
        if f(a) * f(mid) <= 0:
            b = mid
        else:
            a = mid
    return (a + b) / 2

def gauss_solve(A, b):
    # solves Ax = b with partial pivoting
    n = len(A)
    M = [row[:] + [b[i]] for i, row in enumerate(A)]
    for i in range(n):
        p = max(range(i, n), key=lambda r: abs(M[r][i]))
        M[i], M[p] = M[p], M[i]
        for r in range(i + 1, n):
            factor = M[r][i] / M[i][i]
            for c in range(i, n + 1):
                M[r][c] -= factor * M[i][c]
    x = [0] * n
    for i in range(n - 1, -1, -1):
        x[i] = (M[i][n] - sum(M[i][j] * x[j] for j in range(i + 1, n))) / M[i][i]
    return x

def lagrange(points, x):
    total = 0
    for i, (xi, yi) in enumerate(points):
        term = yi
        for j, (xj, _) in enumerate(points):
            if i != j:
                term *= (x - xj) / (xi - xj)
        total += term
    return total

def euler_method(f, x0, y0, x_end, steps):
    # solves dy/dx = f(x, y)
    h = (x_end - x0) / steps
    x, y = x0, y0
    for _ in range(steps):
        y += h * f(x, y)
        x += h
    return y

def crt(remainders, moduli):
    # Chinese remainder theorem (moduli must be pairwise coprime)
    M = math.prod(moduli)
    x = 0
    for r, m in zip(remainders, moduli):
        Mi = M // m
        x += r * Mi * pow(Mi, -1, m)
    return x % M

def linear_regression(xs, ys):
    n = len(xs)
    sx, sy = sum(xs), sum(ys)
    sxy = sum(x * y for x, y in zip(xs, ys))
    sxx = sum(x * x for x in xs)
    slope = (n * sxy - sx * sy) / (n * sxx - sx ** 2)
    return slope, (sy - slope * sx) / n

def dot(a, b):
    return sum(x * y for x, y in zip(a, b))

def cross(a, b):
    return (a[1] * b[2] - a[2] * b[1],
            a[2] * b[0] - a[0] * b[2],
            a[0] * b[1] - a[1] * b[0])

def fib_matrix(n):
    # [[1,1],[1,0]]^n gives Fibonacci numbers
    def mul(A, B):
        return [[sum(A[i][k] * B[k][j] for k in range(2)) for j in range(2)]
                for i in range(2)]
    result, base = [[1, 0], [0, 1]], [[1, 1], [1, 0]]
    while n:
        if n & 1:
            result = mul(result, base)
        base = mul(base, base)
        n >>= 1
    return result[0][1]

def binomial_prob(n, k, p):
    return math.comb(n, k) * p ** k * (1 - p) ** (n - k)

print(f"Root of x^3 - x - 2     : {bisection(lambda x: x**3 - x - 2, 1, 2):.6f}")

A = [[2, 1, -1], [-3, -1, 2], [-2, 1, 2]]
b = [8, -11, -3]
print("Solution of 3 equations :", [round(v, 6) + 0.0 for v in gauss_solve(A, b)])

print("Lagrange f(4) for (1,1),(2,4),(3,9):", round(lagrange([(1, 1), (2, 4), (3, 9)], 4), 6))
print(f"Euler method e^1        : {euler_method(lambda x, y: y, 0, 1, 1, 1000):.6f} | math.e = {math.e:.6f}")

print("Inverse of 3 mod 11     :", pow(3, -1, 11))
print("CRT x=2(3), 3(5), 2(7)  :", crt([2, 3, 2], [3, 5, 7]))

print("Fractions 1/3 + 1/6     :", Fraction(1, 3) + Fraction(1, 6))
print("Fractions 3/4 * 2/9     :", Fraction(3, 4) * Fraction(2, 9))

slope, intercept = linear_regression([1, 2, 3, 4, 5], [2, 4, 5, 4, 5])
print(f"Regression line         : y = {slope:.2f}x + {intercept:.2f}")

a, b = (1, 2, 3), (4, 5, 6)
print("Dot product             :", dot(a, b))
print("Cross product           :", cross(a, b))

print("Fibonacci(10) by matrix :", fib_matrix(10))
print(f"P(3 heads in 10 tosses) : {binomial_prob(10, 3, 0.5):.4f}")
print("255 in binary / hex     :", bin(255)[2:], "/", hex(255)[2:])
```
