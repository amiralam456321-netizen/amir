# “Series”
Fibonacci, arithmetic, and geometric series, plus the sum of the first n terms


```python
def fibonacci(n):
    a, b = 0, 1
    series = []
    for _ in range(n):
        series.append(a)
        a, b = b, a + b
    return series

def arithmetic(first, diff, n):
    return [first + i * diff for i in range(n)]

def geometric(first, ratio, n):
    return [first * ratio ** i for i in range(n)]

n = int(input("How many terms? "))

fib = fibonacci(n)
ap = arithmetic(2, 3, n)    # 2, 5, 8, 11, ...
gp = geometric(2, 2, n)     # 2, 4, 8, 16, ...

print("Fibonacci :", fib, "| Sum =", sum(fib))
print("Arithmetic:", ap, "| Sum =", sum(ap))
print("Geometric :", gp, "| Sum =", sum(gp))
```
