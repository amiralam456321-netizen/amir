Heron’s formula, shoelace polygon area, distance and angle between vectors, complex and polar form, Euler’s identity, the sieve of Eratosthenes, divisor functions, derangements, Roman numerals, digital root, 2x2 matrix inverse, the golden ratio and Ramanujan’s series for pi.

```python
import math
import cmath

def heron(a, b, c):
    s = (a + b + c) / 2
    return math.sqrt(s * (s - a) * (s - b) * (s - c))

def polygon_area(pts):
    # shoelace formula
    n = len(pts)
    total = sum(pts[i][0] * pts[(i + 1) % n][1] - pts[(i + 1) % n][0] * pts[i][1]
                for i in range(n))
    return abs(total) / 2

def distance(p, q):
    return math.dist(p, q)

def angle_between(u, v):
    d = sum(a * b for a, b in zip(u, v))
    return math.degrees(math.acos(d / (math.hypot(*u) * math.hypot(*v))))

def sieve(limit):
    flags = [True] * (limit + 1)
    flags[0:2] = [False, False]
    for i in range(2, int(limit ** 0.5) + 1):
        if flags[i]:
            for j in range(i * i, limit + 1, i):
                flags[j] = False
    return [i for i, f in enumerate(flags) if f]

def divisors(n):
    return [i for i in range(1, n + 1) if n % i == 0]

def derangements(n):
    # number of permutations with no fixed point
    d = [1, 0]
    for i in range(2, n + 1):
        d.append((i - 1) * (d[i - 1] + d[i - 2]))
    return d[:n + 1]

def to_roman(num):
    table = [(1000, "M"), (900, "CM"), (500, "D"), (400, "CD"), (100, "C"),
             (90, "XC"), (50, "L"), (40, "XL"), (10, "X"), (9, "IX"),
             (5, "V"), (4, "IV"), (1, "I")]
    out = ""
    for value, symbol in table:
        while num >= value:
            out += symbol
            num -= value
    return out

def digital_root(n):
    return 0 if n == 0 else 1 + (n - 1) % 9

def inverse_2x2(M):
    (a, b), (c, d) = M
    det = a * d - b * c
    if det == 0:
        raise ValueError("matrix is singular")
    return [[d / det, -b / det], [-c / det, a / det]]

def golden_ratio(iterations=20):
    # continued fraction 1 + 1/(1 + 1/(1 + ...))
    x = 1
    for _ in range(iterations):
        x = 1 + 1 / x
    return x

def ramanujan_pi(terms=2):
    total = sum(math.factorial(4 * k) * (1103 + 26390 * k) /
                (math.factorial(k) ** 4 * 396 ** (4 * k))
                for k in range(terms))
    return 1 / (2 * math.sqrt(2) / 9801 * total)

print("Triangle 3-4-5 area        :", heron(3, 4, 5))
print("Rectangle area (shoelace)  :", polygon_area([(0, 0), (4, 0), (4, 3), (0, 3)]))
print("Distance (1,2) to (4,6)    :", distance((1, 2), (4, 6)))
print(f"Angle between (1,0),(1,1)  : {angle_between((1, 0), (1, 1)):.2f} degrees")

z = 3 + 4j
r, theta = cmath.polar(z)
print(f"|3+4i| = {r:.4f}, angle = {theta:.4f} rad")
e_pi = cmath.exp(1j * math.pi)
print("e^(i*pi)                   :", complex(round(e_pi.real, 6) + 0.0, round(e_pi.imag, 6) + 0.0))

print("Primes up to 30            :", sieve(30))
d = divisors(28)
print("Divisors of 28             :", d, "| count =", len(d), "| sum =", sum(d))
print("Derangements 0..6          :", derangements(6))
print("1994 in Roman numerals     :", to_roman(1994))
print("Digital root of 9875       :", digital_root(9875))
print("Inverse of [[4,7],[2,6]]   :", inverse_2x2([[4, 7], [2, 6]]))
print(f"Golden ratio               : {golden_ratio():.6f} | exact = {(1 + math.sqrt(5)) / 2:.6f}")
print(f"pi by Ramanujan (2 terms)  : {ramanujan_pi():.12f} | math.pi = {math.pi:.12f}")
```
