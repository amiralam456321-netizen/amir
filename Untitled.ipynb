# “Series”
: pentagonal, hexagonal, tetrahedral, Mersenne, Fermat, Stern’s diatomic, Hofstadter Q, Kaprekar, automorphic and abundant numbers, plus the Nilakantha series for pi and the sinh and ln(1+x) series.


```python
import math

def pentagonal(n):
    return [i * (3 * i - 1) // 2 for i in range(1, n + 1)]

def hexagonal(n):
    return [i * (2 * i - 1) for i in range(1, n + 1)]

def tetrahedral(n):
    return [i * (i + 1) * (i + 2) // 6 for i in range(1, n + 1)]

def mersenne(n):
    return [2 ** i - 1 for i in range(1, n + 1)]

def fermat(n):
    return [2 ** (2 ** i) + 1 for i in range(n)]

def stern(n):
    # s(0)=0, s(1)=1, s(2k)=s(k), s(2k+1)=s(k)+s(k+1)
    s = [0, 1]
    for i in range(2, n):
        s.append(s[i // 2] if i % 2 == 0 else s[i // 2] + s[i // 2 + 1])
    return s[:n]

def hofstadter_q(n):
    # Q(i) = Q(i - Q(i-1)) + Q(i - Q(i-2))
    q = [0, 1, 1]
    for i in range(3, n + 1):
        q.append(q[i - q[i - 1]] + q[i - q[i - 2]])
    return q[1:n + 1]

def kaprekar(limit):
    out = []
    for num in range(1, limit + 1):
        sq = str(num * num)
        if num == 1 or any(
            int(sq[i:]) > 0 and int(sq[:i]) + int(sq[i:]) == num
            for i in range(1, len(sq))
        ):
            out.append(num)
    return out

def automorphic(limit):
    # square ends with the number itself
    return [i for i in range(1, limit + 1) if str(i * i).endswith(str(i))]

def abundant(limit):
    # sum of proper divisors is greater than the number
    return [n for n in range(1, limit + 1)
            if sum(i for i in range(1, n // 2 + 1) if n % i == 0) > n]

def nilakantha(terms):
    # pi = 3 + 4/(2*3*4) - 4/(4*5*6) + 4/(6*7*8) - ...
    total = 3
    for k in range(1, terms + 1):
        a = 2 * k
        total += (-1) ** (k + 1) * 4 / (a * (a + 1) * (a + 2))
    return total

def sinh_series(x, terms):
    # sinh(x) = x + x^3/3! + x^5/5! + ...
    return sum(x ** (2 * k + 1) / math.factorial(2 * k + 1)
               for k in range(terms))

def ln_series(x, terms):
    # ln(1+x) = x - x^2/2 + x^3/3 - ...  (-1 < x <= 1)
    return sum((-1) ** (k + 1) * x ** k / k for k in range(1, terms + 1))

n = int(input("How many terms? "))

print("Pentagonal  :", pentagonal(n))
print("Hexagonal   :", hexagonal(n))
print("Tetrahedral :", tetrahedral(n))
print("Mersenne    :", mersenne(n))
print("Fermat      :", fermat(n))
print("Stern       :", stern(n))
print("Hofstadter Q:", hofstadter_q(n))
print("Kaprekar numbers up to 1000 :", kaprekar(1000))
print("Automorphic numbers up to 1000:", automorphic(1000))
print("Abundant numbers up to 50   :", abundant(50))

print(f"\npi by Nilakantha ({n} terms): {nilakantha(n):.6f} | math.pi = {math.pi:.6f}")
print(f"sinh(1) ({n} terms)        : {sinh_series(1, n):.6f} | math.sinh(1) = {math.sinh(1):.6f}")
print(f"ln(1.5) ({n} terms)        : {ln_series(0.5, n):.6f} | math.log(1.5) = {math.log(1.5):.6f}")
```
