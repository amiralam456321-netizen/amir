# “Series”
Lucas, Tribonacci, Leibniz series for pi, cosine series, Armstrong numbers, perfect numbers, and a pandas Series.


```python
import math
import pandas as pd

def lucas(n):
    a, b = 2, 1
    out = []
    for _ in range(n):
        out.append(a)
        a, b = b, a + b
    return out

def tribonacci(n):
    seq = [0, 0, 1]
    while len(seq) < n:
        seq.append(sum(seq[-3:]))
    return seq[:n]

def pi_series(terms):
    # pi = 4 * (1 - 1/3 + 1/5 - 1/7 + ...)
    return 4 * sum((-1) ** k / (2 * k + 1) for k in range(terms))

def cos_series(x, terms):
    # cos(x) = 1 - x^2/2! + x^4/4! - ...
    return sum((-1) ** k * x ** (2 * k) / math.factorial(2 * k)
               for k in range(terms))

def armstrong(limit):
    result = []
    for num in range(1, limit + 1):
        digits = str(num)
        if sum(int(d) ** len(digits) for d in digits) == num:
            result.append(num)
    return result

def perfect(limit):
    return [n for n in range(2, limit + 1)
            if sum(i for i in range(1, n // 2 + 1) if n % i == 0) == n]

n = int(input("How many terms? "))

print("Lucas     :", lucas(n))
print("Tribonacci:", tribonacci(n))
print(f"pi by series ({n} terms): {pi_series(n):.6f}  |  math.pi: {math.pi:.6f}")
print(f"cos(pi/3) by series ({n} terms): {cos_series(math.pi / 3, n):.6f}")
print("Armstrong numbers up to 1000:", armstrong(1000))
print("Perfect numbers up to 10000 :", perfect(10000))

# pandas Series
s = pd.Series([2, 4, 6, 8, 10], name="even_numbers")
print("\npandas Series:")
print(s)
print("Sum:", s.sum(), "| Mean:", s.mean(), "| Max:", s.max())
```
