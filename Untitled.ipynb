**Guessing game**

```python
import random

print("=== Number Guessing Game ===")
print("Main 1 se 100 ke beech ek number soch raha hoon.")

number = random.randint(1, 100)
attempts = 0

while True:
    guess = input("Apna guess likho: ")

    if not guess.isdigit():
        print("Please sirf number likho!")
        continue

    guess = int(guess)
    attempts += 1

    if guess < number:
        print("Zyada bada number likho (Too low)")
    elif guess > number:
        print("Chhota number likho (Too high)")
    else:
        print(f"Shabaash! Aapne {attempts} koshish me sahi number pakda: {number}")
        break
```
