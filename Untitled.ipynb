# “Series”
harmonic, prime numbers, squares and cubes, and the Taylor series for sin(x).


```python
import math

def harmonic(n):
    return [1 / (i + 1) for i in range(n)]          # 1, 1/2, 1/3, ...

def primes(n):
    result, num = [], 2
    while len(result) < n:
        if all(num % p != 0 for p in result if p * p <= num):
            result.append(num)
        num += 1
    return result

def squares(n):
    return [i ** 2 for i in range(1, n + 1)]        # 1, 4, 9, 16, ...

def cubes(n):
    return [i ** 3 for i in range(1, n + 1)]        # 1, 8, 27, 64, ...

def sin_series(x, terms):
    # sin(x) = x - x^3/3! + x^5/5! - ...
    return sum((-1) ** k * x ** (2 * k + 1) / math.factorial(2 * k + 1)
               for k in range(terms))

n = int(input("How many terms? "))

print("Harmonic:", [round(h, 3) for h in harmonic(n)])
print("Primes  :", primes(n))
print("Squares :", squares(n))
print("Cubes   :", cubes(n))

x = math.pi / 6   # 30 degrees
print(f"sin(pi/6) by series ({n} terms): {sin_series(x, n):.6f}")
print(f"sin(pi/6) by math.sin          : {math.sin(x):.6f}")
```
