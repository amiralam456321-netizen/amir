It uses plenty of logic with conditions, loops, and counters.

```python
password = input("Enter a password to check: ")

has_upper = False
has_lower = False
has_digit = False
has_special = False

special_chars = "!@#$%^&*()-_=+[]{};:,.<>?/"

for ch in password:
    if ch.isupper():
        has_upper = True
    elif ch.islower():
        has_lower = True
    elif ch.isdigit():
        has_digit = True
    elif ch in special_chars:
        has_special = True

# Count how many rules are met
score = 0

if len(password) >= 8:
    score += 1
if has_upper:
    score += 1
if has_lower:
    score += 1
if has_digit:
    score += 1
if has_special:
    score += 1

print("\nPassword check:")
print("Length 8+      :", len(password) >= 8)
print("Uppercase      :", has_upper)
print("Lowercase      :", has_lower)
print("Number         :", has_digit)
print("Special symbol :", has_special)

if score == 5:
    print("\nStrength: Strong")
elif score >= 3:
    print("\nStrength: Medium")
else:
    print("\nStrength: Weak")

if not (has_upper and has_digit and has_special):
    print("Tip: add an uppercase letter, a number, and a symbol to make it stronger.")
```
