# “Series”
Sylvester, Recamán, Jacobsthal, Bell numbers, Harshad numbers, twin primes, and the series for e^x, arctan(x) and the Wallis product for pi.


```python
import math

def sylvester(n):
    out, s = [], 2
    for _ in range(n):
        out.append(s)
        s = s * s - s + 1
    return out

def recaman(n):
    seq, seen = [0], {0}
    for i in range(1, n):
        nxt = seq[-1] - i
        if nxt < 0 or nxt in seen:
            nxt = seq[-1] + i
        seq.append(nxt)
        seen.add(nxt)
    return seq

def jacobsthal(n):
    a, b = 0, 1
    out = []
    for _ in range(n):
        out.append(a)
        a, b = b, b + 2 * a
    return out

def bell(n):
    # Bell triangle method
    row, out = [1], []
    for _ in range(n):
        out.append(row[0])
        new = [row[-1]]
        for x in row:
            new.append(new[-1] + x)
        row = new
    return out

def harshad(limit):
    # divisible by the sum of its digits
    return [i for i in range(1, limit + 1)
            if i % sum(int(d) for d in str(i)) == 0]

def twin_primes(limit):
    sieve = [True] * (limit + 1)
    sieve[0:2] = [False, False]
    for i in range(2, int(limit ** 0.5) + 1):
        if sieve[i]:
            for j in range(i * i, limit + 1, i):
                sieve[j] = False
    return [(p, p + 2) for p in range(2, limit - 1)
            if sieve[p] and sieve[p + 2]]

def exp_series(x, terms):
    # e^x = 1 + x + x^2/2! + x^3/3! + ...
    return sum(x ** k / math.factorial(k) for k in range(terms))

def arctan_series(x, terms):
    # arctan(x) = x - x^3/3 + x^5/5 - ...  (|x| <= 1)
    return sum((-1) ** k * x ** (2 * k + 1) / (2 * k + 1)
               for k in range(terms))

def wallis(terms):
    # pi/2 = (2/1 * 2/3) * (4/3 * 4/5) * ...
    prod = 1
    for k in range(1, terms + 1):
        prod *= (4 * k * k) / (4 * k * k - 1)
    return 2 * prod

n = int(input("How many terms? "))

print("Sylvester :", sylvester(n))
print("Recaman   :", recaman(n))
print("Jacobsthal:", jacobsthal(n))
print("Bell      :", bell(n))
print("Harshad numbers up to 50:", harshad(50))
print("Twin primes up to 100   :", twin_primes(100))

print(f"\ne^2 by series ({n} terms): {exp_series(2, n):.6f} | math.exp(2) = {math.exp(2):.6f}")
print(f"arctan(0.5) ({n} terms) : {arctan_series(0.5, n):.6f} | math.atan(0.5) = {math.atan(0.5):.6f}")
print(f"pi by Wallis ({n} terms): {wallis(n):.6f} | math.pi = {math.pi:.6f}")
```
